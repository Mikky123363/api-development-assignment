```python
import requests

response = requests.get("https://api.frankfurter.dev/v1/latest?from=USD")
data = response.json()

print("Base:", data["base"])
print("Date:", data["date"])
print("Number of currencies:", len(data["rates"]))
```

```python
response = requests.get("https://api.frankfurter.dev/v1/currencies")
currencies = response.json()

for code, name in currencies.items():
    print(code, "-", name)

print("\nTotal supported currencies:", len(currencies))
print("Is NGN supported?", "NGN" in currencies)
```

```python
def get_rates(base):
    try:
        response = requests.get(f"https://api.frankfurter.dev/v1/latest?from={base}")
        if response.status_code == 200:
            return response.json()
        else:
            return f"Error: API returned status {response.status_code} - {response.text}"
    except requests.exceptions.RequestException as e:
        return f"Error: request failed - {e}"


test_currencies = ["USD", "EUR", "GBP", "JPY", "CHF", "NGN", "ZWL", "XYZ", "ng"]

for currency in test_currencies:
    result = get_rates(currency)
    print(f"{currency}: {result}\n")
```

```python
response = requests.get("https://jsonplaceholder.typicode.com/users")
users = response.json()

user_dict = {}
for user in users:
    user_dict[user["id"]] = {
        "name": user["name"],
        "email": user["email"],
        "city": user["address"]["city"]
    }

print(user_dict)
```

```python
new_post = {
    "title": "My test post",
    "body": "This is a test post from the assignment.",
    "userId": 1
}

response = requests.post("https://jsonplaceholder.typicode.com/posts", json=new_post)
created = response.json()

print("Status code:", response.status_code)
print("Returned data:", created)

new_id = created["id"]

check_response = requests.get(f"https://jsonplaceholder.typicode.com/posts/{new_id}")
print("\nGET status code:", check_response.status_code)
print("GET result:", check_response.json())
```
